### Use LiteLLM

```python
# anthropic/claude-haiku-4-5
anthropic_response = completion(model="anthropic/claude-haiku-4-5", messages=[{"role": "user", "content": "Write a haiku about soccer!"}])

# openai/gpt-4o-mini
openai_response = completion(model="openai/gpt-4o-mini", messages=[{"role": "user", "content": "Write a haiku about soccer!"}])

# gemini/gemini-3.6-flash
gemini_response = completion(model="gemini/gemini-3.6-flash", messages=[{"role": "user", "content": "Write a haiku about soccer!"}])

# deepseek/deepseek-v4-flash
deepseek_response = completion(model="deepseek/deepseek-v4-flash", messages=[{"role": "user", "content": "Write a haiku about soccer!"}])
```

```python
print("Anthropic response:")
print(anthropic_response.choices[0].message.content)
print('------------------------------------')

print("OpenAI response:")
print(openai_response.choices[0].message.content)

print('------------------------------------')
print("Gemini response:")
print(gemini_response.choices[0].message.content)

print('------------------------------------')
print("DeepSeek response:")
print(deepseek_response.choices[0].message.content)
```

### How it works

LiteLLM acts as a translator between your application and the model provider.

#### Anthropic Models

```python
anthropic_response = completion(
    model="claude-haiku-4-5",
    messages=[
        {
              "role": "user",
              "content": "Write a haiku about soccer!"
        }
      ]
  )
```

```python
import anthropic

client = anthropic.Anthropic(api_key=os.getenv("ANTHROPIC_API_KEY"))

response = client.messages.create(
    model="claude-haiku-4-5",
    max_tokens=1024,
    messages=[
        {
            "role": "user",
            "content": "Write a haiku about soccer!"
        }
    ]
)
```

#### OpenAI Models

```python
open_ai_response = completion(
    model="openai/gpt-4o-mini",
    messages=[
        {
              "role": "user",
              "content": "Write a haiku about soccer!"
        }
      ]
  )
```

```python
from openai import OpenAI

client = OpenAI(api_key=os.getenv("OPENAI_API_KEY"))

response = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {
            "role": "user",
            "content": "Write a haiku about soccer!"
        }
    ]
)
```
