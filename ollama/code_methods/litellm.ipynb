```python
from litellm import completion

# The "ollama_chat/" prefix tells LiteLLM to route the call to your local Ollama server (defaults to localhost:11434)
response = completion(
    model="ollama_chat/llama3.2:1b",
    messages=[{"role": "user", "content": "Why is the sky blue? Answer in one sentence."}],
)

print(response.choices[0].message.content)
```
