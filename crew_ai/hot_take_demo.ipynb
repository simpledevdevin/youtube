# Hot Take Crew — Video Agenda

## 1. Recap: Setup & project creation
- How to install CrewAI (using uv)
- Creating a new crew project: `crewai create crew <project_name>`
- Summarize project layout (important files)

## 2. Recap: Core concepts
- **Agents** — roles, goals, backstories (defined in `agents.yaml`)
- **Tasks** — what each agent does, inputs/outputs, dependencies (defined in `tasks.yaml`)
- How the **Crew** wires agents and tasks into a runnable workflow

## 3. Summarize this project’s workflow (Hot Take)
- **Moderator** — chooses the debate topic (e.g. “remote work”, “AI in education”).
- **Junior Debater** — proposes an initial hot take on that topic.
- **Senior Debater** — refines and strengthens the junior debater’s argument.
- **Counter Debater** — argues against the hot take and plays devil’s advocate.


## 4. Code walkthrough
- **agents.yaml** — role definitions, goals, backstories for Moderator, Junior/Senior/Counter Debater.
- **tasks.yaml** — task descriptions, expected outputs, and task ordering/dependencies.
    - highlight the "context" parameter. Mention that it could be used in crew.py instead if wanted
- **crew.py** — instantiates agents and tasks, builds the crew, runs the kickoff.
    - highlight use of tool SerperDevTool 
- **main.py** — entry point (CLI/args), calls the crew, handles inputs/outputs.
- **.env** — API keys (e.g. `OPENAI_API_KEY`) — remind viewers not to commit secrets.
