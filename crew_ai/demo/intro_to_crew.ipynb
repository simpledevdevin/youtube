## 1. What is CrewAI?

- CrewAI is an open-source framework that lets you build teams of AI agents that work together to solve problems.
   - Simple structure: agents, tasks, and crews.
- Key ideas: **agents** (roles with goals and backstories) and **tasks** (steps that use those agents and pass outputs between them).
- Brief, high-level explanation before diving into setup.
- https://docs.crewai.com/

## 3. Setup guide

1. **Install [uv](https://github.com/astral-sh/uv)** (fast Python package manager):
   - Used by the CrewAI CLI for environments and dependencies.
2. **Install the CrewAI CLI:**
   - e.g. `uv tool install crewai"`
3. **Verify tools:**
   - Run `uv tool list` to confirm uv and crewai (or relevant) tools are available.

---

## 4. Create a CrewAI project

- **Command:**  
  `crewai create crew <your_project_name>`
- Walk through what this generates: project layout, config files, and entry points.

### Project layout (what to explain)

| File / folder   | Purpose |
|-----------------|--------|
| `agents.yaml`   | Agent definitions: role, goal, backstory. |
| `tasks.yaml`    | Task definitions: description, assigned agent, inputs/outputs. |
| `crew.py`       | Assembles agents and tasks into a crew; runs the workflow. |
| `main.py`       | Entry point (CLI, args); invokes the crew and handles I/O. |
| `.env`          | API keys (e.g. `OPENAI_API_KEY`). **Do not commit.** |
