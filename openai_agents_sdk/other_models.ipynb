### Define OpenAI Compatible Endpoints

Set the base URLs for the Gemini and DeepSeek clients.

```python
GEMINI_BASE_URL = "https://generativelanguage.googleapis.com/v1beta/openai/"
DEEPSEEK_BASE_URL = "https://api.deepseek.com/v1"
```

| Provider          | Base URL                                                   |
| ----------------- | ---------------------------------------------------------- |
| Google Gemini     | `https://generativelanguage.googleapis.com/v1beta/openai/` |
| Groq              | `https://api.groq.com/openai/v1`                           |
| Together AI       | `https://api.together.ai/v1`                               |
| OpenRouter        | `https://openrouter.ai/api/v1`                             |
| Cerebras          | `https://api.cerebras.ai/v1`                               |
| Ollama (local)    | `http://localhost:11434/v1`                                |
| LM Studio (local) | `http://localhost:1234/v1`                                 |
| DeepInfra         | `https://api.deepinfra.com/v1/openai`                      |
| Perplexity        | `https://api.perplexity.ai`                                |
| Mistral           | `https://api.mistral.ai/v1`                                |
| xAI (Grok)        | `https://api.x.ai/v1`                                      |


The first step is creating a client. The client simply knows how to connect to a specific AI provider, such as Gemini or DeepSeek, using the correct endpoint and API key. We do this with `AsyncOpenAI`

Think of a client as a messenger between your code and the AI provider. When your code wants a response from Gemini or DeepSeek, the client handles sending the request and getting the answer back.

```python
gemini_client = AsyncOpenAI(base_url=GEMINI_BASE_URL, api_key=google_api_key)
deepseek_client = AsyncOpenAI(base_url=DEEPSEEK_BASE_URL, api_key=deepseek_api_key)
```

Next, we create a model object using that client. This allows the Agents SDK to use Gemini or DeepSeek just like it would an OpenAI model. We do this with `OpenAIChatCompletionsModel`. 

```python
gemini_model = OpenAIChatCompletionsModel(model="gemini-2.5-flash", openai_client=gemini_client)
deepseek_model = OpenAIChatCompletionsModel(model="deepseek-v4-flash", openai_client=deepseek_client)
```

### Gemini

Run the same prompt with the Gemini model.

```python
poet = Agent(
  name="gemini_poet",
  instructions="You are poet who creates beautiful poetry",
  model=gemini_model)

result = await Runner.run(poet, "Write a haiku about artificial intelligence.")
# Runner.run_sync -  blocking/synchronous version
print(result.final_output)
```

### DeepSeek

Run the same prompt with the DeepSeek model.

```python
poet = Agent(
  name="deepseek_poet",
  instructions="You are poet who creates beautiful poetry",
  model=deepseek_model)

result = await Runner.run(poet, "Write a haiku about artificial intelligence.")
# Runner.run_sync -  blocking/synchronous version
print(result.final_output)
```
